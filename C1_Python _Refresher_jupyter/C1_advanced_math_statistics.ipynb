## Advanced Math Functions

```python
import math
import numpy as np

# Solving a quadratic equation: ax^2 + bx + c = 0
a, b, c = 1, -3, 2
D = b**2 - 4*a*c
root1 = (-b + math.sqrt(D)) / (2*a)
root2 = (-b - math.sqrt(D)) / (2*a)
print(f"Roots: {root1}, {root2}")

# Derivative approximation
def f(x):
    return x**3 + 2*x

x0 = 1.0
h = 1e-5
numerical_derivative = (f(x0 + h) - f(x0)) / h
print(f"Approximate derivative at x={x0}: {numerical_derivative}")

# Integration approximation
integral_approximation = np.trapz([f(x) for x in np.linspace(0, 1, 100)], dx=0.01)
print(f"Approximate integral from 0 to 1: {integral_approximation}")
```

## Probability and Statistics

```python
import scipy.stats as stats

# Normal distribution
mean = 0
std_dev = 1
x = np.linspace(-3, 3, 100)
pdf = stats.norm.pdf(x, mean, std_dev)
cdf = stats.norm.cdf(x, mean, std_dev)

print(f"PDF at x=0: {stats.norm.pdf(0, mean, std_dev)}")
print(f"CDF at x=0: {stats.norm.cdf(0, mean, std_dev)}")

# Random sampling
samples = np.random.normal(loc=0, scale=1, size=1000)
```

## Descriptive Statistics

```python
mean_value = np.mean(samples)
median_value = np.median(samples)
mode_value = stats.mode(samples, keepdims=False).mode
print(f"Mean: {mean_value}, Median: {median_value}, Mode: {mode_value}")

variance = np.var(samples)
std_dev = np.std(samples)
print(f"Variance: {variance}, Standard Deviation: {std_dev}")
```

## Inferential Statistics

```python
# One-sample t-test
population_mean = 0
t_statistic, p_value = stats.ttest_1samp(samples, population_mean)
print(f"t-statistic: {t_statistic}, p-value: {p_value}")

# Confidence Interval
conf_interval = stats.norm.interval(0.95, loc=mean_value, scale=std_dev/np.sqrt(len(samples)))
print(f"95% Confidence Interval for the mean: {conf_interval}")
```

## Linear Algebra

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 2]])

# Matrix multiplication
product = np.matmul(A, B)
print(f"Matrix Product:\n{product}")

# Inverse of a matrix
inverse_A = np.linalg.inv(A)
print(f"Inverse of A:\n{inverse_A}")

# Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)
print(f"Eigenvalues: {eigenvalues}\nEigenvectors:\n{eigenvectors}")
```

## Machine Learning Foundations

```python
from sklearn.linear_model import LinearRegression

# Dummy data
X = np.random.rand(100, 1)
y = 3*X.squeeze() + 2 + np.random.randn(100)*0.1

model = LinearRegression()
model.fit(X, y)

print(f"Intercept: {model.intercept_}")
print(f"Coefficient: {model.coef_[0]}")

# Prediction
prediction = model.predict(np.array([[0.5]]))
print(f"Prediction for X=0.5: {prediction[0]}")
```
