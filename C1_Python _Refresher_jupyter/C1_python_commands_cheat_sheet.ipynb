# Python Commands Cheat Sheet Notebook

This notebook contains a curated list of the most important Python commands, categorized by functionality.
Use it as a quick reference guide for everyday Python programming and coding interviews.

## Basic Commands

```python
print("Hello, World!")
# This is a comment
x = 5
y = "Python"
```

## Data Types

```python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
my_set = {1, 2, 3}
my_dict = {"name": "Alice", "age": 25}
```

## Conditional Statements

```python
x = 10
if x > 5:
    print("x is greater than 5")
elif x == 5:
    print("x is 5")
else:
    print("x is less than 5")
```

## Loops

```python
for i in range(5):
    print(i)

count = 0
while count < 5:
    print(count)
    count += 1
```

## Functions

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("World"))

square = lambda x: x**2
print(square(4))
```

## Exception Handling

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero.")
finally:
    print("Execution completed.")
```

## Working with Files

```python
with open("example.txt", "w") as f:
    f.write("Hello, file!")

with open("example.txt", "r") as f:
    content = f.read()
    print(content)
```

## Useful Built-in Functions

```python
print(len(my_list))
print(type(x))
print(sorted([3, 1, 2]))
print(sum([1, 2, 3]))
```

## Importing Modules

```python
import math
print(math.sqrt(16))

import random
print(random.randint(1, 10))
```

## List Comprehensions

```python
squares = [x**2 for x in range(5)]
print(squares)
```

## Dictionary Comprehensions

```python
squared_numbers = {x: x**2 for x in range(5)}
print(squared_numbers)
```

## Working with NumPy

```python
import numpy as np

arr = np.array([1, 2, 3, 4])
print(arr * 2)
```

## Working with pandas

```python
import pandas as pd

df = pd.DataFrame({"Name": ["Alice", "Bob"], "Age": [25, 30]})
print(df)
```
