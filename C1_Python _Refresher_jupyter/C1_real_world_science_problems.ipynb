This notebook demonstrates how Python solves real-world scientific problems.
Each example includes a simple explanation followed by working code.

## 1. Population Growth Prediction (Biology)

**Explanation:**

Biologists predict animal population growth using simple math. Python can help model this.

```python
initial_population = 100
growth_rate = 0.05  # 5% per year
years = 10

future_population = initial_population * (1 + growth_rate) ** years
print(f"Estimated population after {years} years: {future_population:.0f} animals")
```

## 2. Radioactive Decay (Physics)

**Explanation:**

Radioactive materials lose mass over time. We can model how much remains.

```python
half_life = 5  # years
initial_amount = 1000  # grams
elapsed_time = 10

remaining_amount = initial_amount * (0.5) ** (elapsed_time / half_life)
print(f"Remaining radioactive material after {elapsed_time} years: {remaining_amount:.1f} grams")
```

## 3. Medicine Dosage Decay (Health Sciences)

**Explanation:**

Medicine amounts in the body reduce over time after a dose.

```python
initial_dose = 500  # mg
half_life_hours = 4
hours_passed = 8

remaining_dose = initial_dose * (0.5) ** (hours_passed / half_life_hours)
print(f"Remaining medicine after {hours_passed} hours: {remaining_dose:.1f} mg")
```

## 4. Climate Change: Average Temperature Rise (Environmental Science)

**Explanation:**

Scientists estimate global temperatures increasing each year.

```python
current_temp = 14.9  # Celsius
temp_increase_per_year = 0.02
projection_years = 50

future_temp = current_temp + (temp_increase_per_year * projection_years)
print(f"Projected global average temperature in {projection_years} years: {future_temp:.2f} °C")
```

## 5. Chemical Reaction Rate (Chemistry)

**Explanation:**

Chemical reactions slow down as reactants are used up.

```python
import numpy as np

initial_concentration = 1.0  # mol/L
rate_constant = 0.1  # per second
time_seconds = 20

concentration = initial_concentration * np.exp(-rate_constant * time_seconds)
print(f"Concentration after {time_seconds} seconds: {concentration:.3f} mol/L")
```

## 6. Simple Genetic Probability (Biology)

**Explanation:**

We can predict a child's gene combinations based on their parents' genes.

```python
parent1_gene = "Aa"
parent2_gene = "Aa"

possible_offspring = [a+b for a in parent1_gene for b in parent2_gene]
print(f"Possible gene combinations: {possible_offspring}")

prob_aa = possible_offspring.count("aa") / len(possible_offspring)
print(f"Probability of homozygous recessive (aa): {prob_aa*100:.1f}%")
```

## 7. Astronomical Distance Conversion (Astronomy)

**Explanation:**

Space distances are so big that we measure them in light-years instead of kilometers.

```python
kilometers_per_light_year = 9.461e12
light_years = 4.24  # Distance to Proxima Centauri

distance_km = light_years * kilometers_per_light_year
print(f"Distance to Proxima Centauri: {distance_km:.2e} kilometers")
```
