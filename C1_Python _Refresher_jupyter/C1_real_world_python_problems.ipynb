## 1. Weather Forecasting

**Explanation:**

We can predict tomorrow's temperature using the average of past temperatures, just like a simple weather app might.

```python
past_temperatures = [20, 21, 19, 22, 20]
predicted_temperature = sum(past_temperatures) / len(past_temperatures)
print(f"Predicted temperature for tomorrow: {predicted_temperature:.1f}°C")
```

## 2. Loan Approval

**Explanation:**

Banks decide if you qualify for a loan based on your income and credit score.

```python
income = 50000
credit_score = 700

if income > 40000 and credit_score > 650:
    print("Loan Approved!")
else:
    print("Loan Denied.")
```

## 3. Spam Email Detection

**Explanation:**

Email systems spot spam by checking for words like 'win' or 'prize'.

```python
email = "Congratulations! You have won a prize! Claim your money now!"
spam_keywords = ["win", "prize", "money", "claim"]

if any(word in email.lower() for word in spam_keywords):
    print("This email is likely spam.")
else:
    print("This email looks safe.")
```

## 4. Budget Planning

**Explanation:**

We make sure our expenses don't exceed our income to stay financially healthy.

```python
monthly_income = 3000
monthly_expenses = {"rent": 1200, "groceries": 400, "transport": 200, "entertainment": 300}

total_expenses = sum(monthly_expenses.values())

if total_expenses > monthly_income:
    print("You are overspending! Time to cut down expenses.")
else:
    print("Your budget looks good.")
```

## 5. Fitness Tracker

**Explanation:**

Fitness apps cheer you on if you reach your daily step goal.

```python
steps_today = 8500
step_goal = 10000

if steps_today >= step_goal:
    print("Congratulations! You've met your step goal.")
else:
    print(f"You have {step_goal - steps_today} steps left to reach your goal.")
```

## 6. Grocery Shopping List Reminder

**Explanation:**

Python can remind us what we forgot to buy during grocery shopping.

```python
shopping_list = ["milk", "bread", "eggs", "apples"]
purchased_items = ["milk", "bread"]

missing_items = set(shopping_list) - set(purchased_items)
print(f"You still need to buy: {', '.join(missing_items)}")
```

## 7. Movie Recommendation

**Explanation:**

Streaming services suggest movies based on your favorite genres.

```python
import numpy as np

movies = {"Action": ["Die Hard", "Mad Max"], "Comedy": ["The Mask", "Superbad"], "Drama": ["Forrest Gump", "The Shawshank Redemption"]}

preferred_genre = "Comedy"
recommended_movie = np.random.choice(movies[preferred_genre])

print(f"Recommended movie for you: {recommended_movie}")
```
