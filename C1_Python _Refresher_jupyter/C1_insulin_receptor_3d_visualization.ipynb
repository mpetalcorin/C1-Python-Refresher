# 3D Visualization of Insulin Bound to Insulin Receptor

This notebook shows how to visualize the 3D molecular structure of insulin bound to its receptor.
Each step includes a simple explanation and beautiful graphics.

## Setup: Install Required Libraries

We will use **py3Dmol**, a powerful 3D viewer inside Jupyter Notebooks.

```python
# Uncomment if py3Dmol is not installed
# !pip install py3Dmol

import py3Dmol
```

## What is Insulin and Its Receptor? 

**Explanation:**

Insulin is a hormone that helps control blood sugar. It acts like a **key** that unlocks a **door** (the insulin receptor) on cells, allowing sugar to enter. Scientists have captured a snapshot of insulin binding to its receptor. Let's visualize it!

## Load 3D Structure (PDB ID: 6CEB)

We fetch the real 3D structure of insulin bound to its receptor.

```python
# Load structure from PDB
viewer = py3Dmol.view(query='pdb:6CEB')
viewer.setStyle({'cartoon': {'color': 'spectrum'}})
viewer.setBackgroundColor('white')
viewer.zoomTo()
viewer.show()
```

## Highlight Insulin and Receptor Separately

Let's color **insulin** and **receptor** differently to better understand the interaction.

```python
# Color chains differently
viewer = py3Dmol.view(query='pdb:6CEB')

viewer.setStyle({'chain': 'A'}, {'cartoon': {'color': 'magenta'}})  # Insulin
viewer.setStyle({'chain': 'E'}, {'cartoon': {'color': 'cyan'}})      # Receptor
viewer.setStyle({'not': {'chain': ['A', 'E']}}, {'cartoon': {'color': 'lightgrey'}})

viewer.setBackgroundColor('white')
viewer.zoomTo()
viewer.show()
```

## Simple Takeaways

- **Purple (Magenta)** represents the insulin molecule.
- **Cyan** represents the insulin receptor fragment.
- The docking (hugging) action triggers sugar uptake from bloodstream into the cells through GLUT transporters of the body with implications in diabetes.

Visualizing structures like this helps scientists design better treatments for diabetes!
