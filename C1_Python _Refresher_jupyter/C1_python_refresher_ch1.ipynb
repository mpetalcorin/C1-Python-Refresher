## Python Basics: Data Types
Understanding integers, floats, strings, and booleans.

```python
my_int = 10
my_float = 3.14
my_str = "Hello, Python!"
my_bool = True

print(type(my_int), type(my_float), type(my_str), type(my_bool))
```

## Lists, Tuples, and Dictionaries
How to organize data in different container types.

```python
fruits = ["apple", "banana", "cherry"]
print(f"Second fruit: {fruits[1]}")

coordinates = (40.7128, -74.0060)
print(f"Coordinates: {coordinates}")

country_codes = {"US": "United States", "FR": "France"}
print(f"US stands for: {country_codes['US']}")
```

## Functions, Loops, and Conditionals
Using functions for reuse and logic to control flow.

```python
def celsius_to_fahrenheit(c):
    return c * 9/5 + 32

print(f"25°C = {celsius_to_fahrenheit(25)}°F")

for i in range(1, 6):
    print(i)

num = -5
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

## NumPy Basics
Efficient numerical operations using arrays.

```python
import numpy as np

array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

print("Element-wise multiplication:", array1 * array2)
```

## pandas Basics
Working with structured data using Series and DataFrames.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35]
}

df = pd.DataFrame(data)
print(df)

# Filtering example
print("People older than 28:")
print(df[df["Age"] > 28])
```
