## Introduction to Data Types

## Writing Functions in Python

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("World"))

name = "Python"
print(name.upper())

def add(x, y):
    return x + y

print(add(5, 3))

def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(5))
```

## Writing Loops and Conditionals

```python
x = 10
if x > 5:
    print("x is greater than 5")

count = 0
while count < 3:
    print(count)
    count += 1

for fruit in fruits:
    print(fruit)

for key, value in person.items():
    print(key, value)
```

## Introduction to NumPy

```python
import numpy as np

array = np.array([1, 2, 3, 4])
print(array)

matrix = np.array([[1, 2], [3, 4]])
print(matrix)

for element in array:
    print(element)
```

## Introduction to pandas

```python
import pandas as pd

data = {"Name": ["Alice", "Bob"], "Age": [25, 30]}
df = pd.DataFrame(data)
print(df)

print(df[df["Age"] > 25])

print(df["Age"].mean())

df["Country"] = ["USA", "Canada"]
print(df)

df.loc[0, "Age"] = 26
print(df)

data2 = {"Name": ["Charlie"], "Age": [35], "Country": ["UK"]}
df2 = pd.DataFrame(data2)
combined_df = pd.concat([df, df2], ignore_index=True)
print(combined_df)

grouped = combined_df.groupby("Country")["Age"].mean()
print(grouped)
```
