# Python Drug Discovery Problems with Explanations and Graphics

This notebook shows how Python solves real-world drug discovery problems.
Each example includes a simple explanation and beautiful graphics to visualize the concepts.

## Setup: Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
```

## 1. Comparing Molecules (Similarity)

**Explanation:**

Finding similar molecules is like finding molecular twins based on size, shape, or features.

```python
# Fake molecule descriptors
molecule_A = [0.8, 0.6, 0.9]
molecule_B = [0.7, 0.65, 0.88]

# Cosine similarity
similarity = np.dot(molecule_A, molecule_B) / (np.linalg.norm(molecule_A) * np.linalg.norm(molecule_B))
print(f"Similarity between molecule A and B: {similarity:.2f}")
```

## 2. Predicting Bioactivity (Regression)

**Explanation:**

Predicting how strongly a molecule binds to its target helps save a lot of lab work.

```python
# Generate fake data
np.random.seed(0)
features = np.random.rand(100)
bioactivity = 3 * features + np.random.normal(0, 0.1, 100)

# Plot
plt.figure(figsize=(8,5))
sns.scatterplot(x=features, y=bioactivity)
plt.title('Molecule Feature vs Bioactivity')
plt.xlabel('Feature Value')
plt.ylabel('Bioactivity Score')
plt.show()
```

## 3. Estimating Drug-Likeness (Rule of Five)

**Explanation:**

A good drug usually isn't too big or too greasy, based on "Lipinski's Rule of Five."

```python
# Example molecule properties
molecular_weight = 450
octanol_water_partition = 4

if molecular_weight < 500 and octanol_water_partition < 5:
    print("This molecule is likely drug-like!")
else:
    print("This molecule might have issues as a drug.")
```

## 4. Machine Learning to Speed Up Discovery

**Explanation:**

Python models predict which molecules to test first, saving time and money.

```python
import warnings
warnings.filterwarnings("ignore")

# Simulate model predictions
molecules = np.arange(1, 21)
predicted_scores = np.random.rand(20)

# Plot
plt.figure(figsize=(10,6))
sns.barplot(x=molecules, y=predicted_scores, palette='viridis')
plt.title('Predicted Molecule Effectiveness')
plt.xlabel('Molecule ID')
plt.ylabel('Predicted Effectiveness Score')
plt.xticks(rotation=45)
plt.show()
```

## 5. Visualizing Molecular Space (PCA Projection)

**Explanation:**

We can plot molecules in 2D space to see clusters and patterns.

```python
# Fake high-dimensional molecule features
high_dim_features = np.random.rand(50, 5)

# Reduce to 2D
pca = PCA(n_components=2)
reduced_features = pca.fit_transform(high_dim_features)

# Plot
plt.figure(figsize=(8,6))
sns.scatterplot(x=reduced_features[:,0], y=reduced_features[:,1], s=100)
plt.title('Molecular Space (PCA 2D Projection)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```
