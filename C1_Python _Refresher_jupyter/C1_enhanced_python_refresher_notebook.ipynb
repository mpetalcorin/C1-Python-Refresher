## Introduction to Data Types

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[0])
print(fruits[-1])
fruits[1] = "blueberry"
print(fruits)

# List Comprehension
squares = [x**2 for x in range(10)]
print(squares)
```

## Tuples and Dictionaries

```python
coordinates = (10, 20)
print(coordinates)

person = {"name": "Alice", "age": 30}
print(person["name"])

# Dictionary Comprehension
squared_numbers = {x: x**2 for x in range(5)}
print(squared_numbers)
```

## Writing Functions in Python

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("World"))

name = "Python"
print(name.upper())

def add(x, y):
    return x + y

print(add(5, 3))

def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(5))

# Lambda Function
square = lambda x: x**2
print(square(5))
```

## Writing Loops and Conditionals

```python
x = 10
if x > 5:
    print("x is greater than 5")

count = 0
while count < 3:
    print(count)
    count += 1

for fruit in fruits:
    print(fruit)

for key, value in person.items():
    print(f"{key}: {value}")

# Enumerate and Zip
for idx, fruit in enumerate(fruits):
    print(f"{idx}: {fruit}")

names = ["Alice", "Bob"]
ages = [25, 30]
for name, age in zip(names, ages):
    print(f"{name} is {age} years old")
```

## Introduction to NumPy

```python
import numpy as np

array = np.array([1, 2, 3, 4])
print(array)

matrix = np.array([[1, 2], [3, 4]])
print(matrix)

# Array Operations
print(array * 2)

# Boolean Masking
mask = array > 2
print(array[mask])
```

## Introduction to pandas

```python
import pandas as pd

# Creating DataFrame
data = {"Name": ["Alice", "Bob"], "Age": [25, 30]}
df = pd.DataFrame(data)
print(df)

# Filtering DataFrame
print(df[df["Age"] > 25])

# Mean of Age
print(df["Age"].mean())

# Add Column
df["Country"] = ["USA", "Canada"]
print(df)

# Setting data with loc[]
df.loc[0, "Age"] = 26
print(df)

# Combining DataFrames
data2 = {"Name": ["Charlie"], "Age": [35], "Country": ["UK"]}
df2 = pd.DataFrame(data2)
combined_df = pd.concat([df, df2], ignore_index=True)
print(combined_df)

# Grouping and Aggregating
grouped = combined_df.groupby("Country")["Age"].mean()
print(grouped)
```

## Bonus: Best Practices and Extras

```python
# F-Strings
name = "Alice"
age = 30
print(f"{name} is {age} years old.")

# List Unpacking
first, second, third = fruits
print(first, second, third)

# Type Hinting
def greet_user(name: str) -> str:
    return f"Hello, {name}!"

print(greet_user("Developer"))

# Try/Except Error Handling
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero.")
```
